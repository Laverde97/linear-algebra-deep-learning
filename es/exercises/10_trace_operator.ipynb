## Traza = suma de la diagonal

```python
A=np.array([[2.,1.,0.],[4.,3.,5.],[1.,2.,7.]])
print("Traza:",np.trace(A))
print("Suma diagonal:",np.diag(A).sum())
```

## Identidad con la norma de Frobenius

```python
left=np.linalg.norm(A,'fro')
right=np.sqrt(np.trace(A@A.T))
print(left,right,np.allclose(left,right))
```

## Invariancia cíclica

```python
B=np.random.randn(3,4); C=np.random.randn(4,3)
print("Tr(BC)=",np.trace(B@C))
print("Tr(CB)=",np.trace(C@B))
```

### Conexión con ML

La traza permite escribir expresiones de varianza, covarianza y normas de manera compacta, algo frecuente en derivaciones de optimización y probabilidades.
