## Matriz identidad

Es el “1” de las matrices: multiplicar por ella deja el objeto igual.

```python
A = np.array([[2.,1.],[5.,3.]])
I = np.eye(2)
print(I)
print("A @ I =")
print(A @ I)
```

## Inversa

Si existe `A⁻¹`, entonces `A⁻¹A = I`.

```python
A_inv = np.linalg.inv(A)
print("A^-1 =")
print(A_inv)
print("A^-1 @ A =")
print(A_inv @ A)
```

## Resolver Ax=b

En código numérico es preferible `solve` a calcular la inversa explícitamente.

```python
b = np.array([5.,13.])
x = np.linalg.solve(A,b)
print("x =", x)
print("Verificación A@x =", A@x)
```

## ¿Qué pasa si no hay inversa?

```python
S = np.array([[1.,2.],[2.,4.]])
print("det(S)=", np.linalg.det(S))
try:
    print(np.linalg.inv(S))
except np.linalg.LinAlgError as e:
    print("No es invertible:", e)
```
