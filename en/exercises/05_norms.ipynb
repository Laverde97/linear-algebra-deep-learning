## Measuring the ‘size’ of a vector

```python
x = np.array([3.,-4.,2.])
print("L1:", np.linalg.norm(x,1))
print("L2:", np.linalg.norm(x,2))
print("L∞:", np.linalg.norm(x,np.inf))
```

## Frobenius norm for matrices

```python
A = np.array([[1.,2.],[3.,4.]])
print("Frobenius:", np.linalg.norm(A,'fro'))
print("sqrt(sum squares):", np.sqrt(np.sum(A**2)))
```

## Distance between two MNIST images

```python
from tensorflow.keras.datasets import mnist
(x_train,y_train),_ = mnist.load_data()
a=x_train[0].astype(float)/255
b=x_train[1].astype(float)/255
print("Etiquetas:", y_train[0], y_train[1])
print("Distancia L2:", np.linalg.norm((a-b).ravel(),2))
print("Distancia L1:", np.linalg.norm((a-b).ravel(),1))
```

```python
plt.figure(figsize=(6,3))
plt.subplot(1,2,1); plt.imshow(a,cmap='gray'); plt.title(str(y_train[0])); plt.axis('off')
plt.subplot(1,2,2); plt.imshow(b,cmap='gray'); plt.title(str(y_train[1])); plt.axis('off')
plt.show()
```

## Regularization: intuition

L1 tends to encourage many weights to be exactly small/zero; L2 penalizes large weights smoothly.

```python
weights = np.array([0.1, -0.2, 2.5, -3.0, 0.05])
print("L1 penalty:", np.linalg.norm(weights,1))
print("L2² penalty:", np.linalg.norm(weights,2)**2)
```
