```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
print("NumPy:", np.__version__)
```

## Identity matrix

It is the ‘1’ of matrices: multiplying by it leaves the object unchanged.

```python
A = np.array([[2.,1.],[5.,3.]])
I = np.eye(2)
print(I)
print("A @ I =")
print(A @ I)
```

## Inverse

Si existe `A⁻¹`, entonces `A⁻¹A = I`.

```python
A_inv = np.linalg.inv(A)
print("A^-1 =")
print(A_inv)
print("A^-1 @ A =")
print(A_inv @ A)
```

## Solving Ax=b

In numerical code, `solve` is preferable to computing the inverse explicitly.

```python
b = np.array([5.,13.])
x = np.linalg.solve(A,b)
print("x =", x)
print("Check A@x =", A@x)
```

## What if there is no inverse?

```python
S = np.array([[1.,2.],[2.,4.]])
print("det(S)=", np.linalg.det(S))
try:
    print(np.linalg.inv(S))
except np.linalg.LinAlgError as e:
    print("Not invertible:", e)
```
