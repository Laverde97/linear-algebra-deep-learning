```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
print("NumPy:", np.__version__)
```

## Factoring a matrix

SVD writes a matrix as `A = U Σ Vᵀ` and also works for non-square matrices.

```python
A=np.array([[3.,1.],[2.,2.],[1.,3.]])
U,s,Vt=np.linalg.svd(A,full_matrices=False)
print("U",U.shape,"s",s.shape,"Vt",Vt.shape)
A_rec=U@np.diag(s)@Vt
print("Correct reconstruction:",np.allclose(A,A_rec))
```

## Compression with MNIST

```python
from tensorflow.keras.datasets import mnist
(x_train,y_train),_=mnist.load_data()
img=x_train[0].astype(float)/255
U,s,Vt=np.linalg.svd(img,full_matrices=False)
print("Image:",img.shape,"singular values:",s.shape)
```

```python
ks=[1,3,5,10,20]
plt.figure(figsize=(15,3))
for i,k in enumerate(ks):
    rec=(U[:,:k]*s[:k])@Vt[:k,:]
    plt.subplot(1,len(ks),i+1); plt.imshow(rec,cmap='gray',vmin=0,vmax=1); plt.title(f'k={k}'); plt.axis('off')
plt.show()
```

## Reconstruction error

```python
for k in [1,3,5,10,20,28]:
    rec=(U[:,:k]*s[:k])@Vt[:k,:]
    err=np.linalg.norm(img-rec,'fro')
    print(f"k={k:2d} | error={err:.4f}")
```

### Connection to AI

SVD enables low-rank approximations, compression, and structural analysis. Similar concepts appear in modern model compression and adaptation techniques.
