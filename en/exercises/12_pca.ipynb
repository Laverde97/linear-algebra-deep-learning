**Goal:** understand the concept, see it with numbers, and connect it to AI/deep learning. Run the cells in order.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
print("NumPy:", np.__version__)
```

## PCA: compressing while preserving as much variation as possible

```python
from tensorflow.keras.datasets import mnist
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
(x_train,y_train),_=mnist.load_data()
X=x_train[:5000].reshape(5000,-1).astype('float32')/255.0
y=y_train[:5000]
print(X.shape)
```

## 784 dimensions → 2 dimensions

```python
pca2=PCA(n_components=2,random_state=42)
Z=pca2.fit_transform(X)
print("Reducido:",Z.shape)
print("Varianza explicada total:",pca2.explained_variance_ratio_.sum())
```

```python
plt.figure(figsize=(8,6))
sc=plt.scatter(Z[:,0],Z[:,1],c=y,s=8,alpha=.6)
plt.colorbar(sc,label='digit')
plt.title('MNIST proyectado a 2 componentes principales')
plt.show()
```

## Reconstruction with 20 components

```python
pca20=PCA(n_components=20,random_state=42)
Z20=pca20.fit_transform(X)
Xrec=pca20.inverse_transform(Z20)
print("Compression:",X.shape[1],"→",Z20.shape[1])
print("Varianza explicada:",pca20.explained_variance_ratio_.sum())
```

```python
plt.figure(figsize=(10,4))
for i in range(5):
    plt.subplot(2,5,i+1); plt.imshow(X[i].reshape(28,28),cmap='gray'); plt.axis('off')
    plt.subplot(2,5,i+6); plt.imshow(Xrec[i].reshape(28,28),cmap='gray'); plt.axis('off')
plt.suptitle('Top: original | Bottom: PCA reconstruction (20 components)')
plt.show()
```

## How many components do I need?

```python
pca=PCA().fit(X)
cum=np.cumsum(pca.explained_variance_ratio_)
for threshold in [0.80,0.90,0.95]:
    k=np.searchsorted(cum,threshold)+1
    print(f"{threshold:.0%} de varianza → {k} componentes")
plt.figure(figsize=(7,4)); plt.plot(cum); plt.axhline(.95,linestyle='--'); plt.xlabel('n componentes'); plt.ylabel('varianza acumulada'); plt.show()
```

### Connection to AI

PCA demonstrates **compact representation**: replacing hundreds of variables with fewer informative directions. It is a conceptual gateway to embeddings and latent representations.
