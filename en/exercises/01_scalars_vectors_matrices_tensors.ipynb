## 1. Simulated random data

We begin with tabular information. A single value is a **scalar**, one row is a **vector**, and the whole table is a **matrix**.

```python
ages = np.random.randint(18, 80, 100)
weight = np.random.uniform(50, 100, 100)
height = np.random.uniform(1.50, 1.95, 100)
income = np.random.uniform(1_000_000, 12_000_000, 100)

data = np.column_stack([ages, weight, height, income])
print("Scalar:", data[0,0])
print("Vector (one person):", data[0].shape)
print("Matrix (100 people):", data.shape)
print(data[:3])
```

## 2. MNIST: an image is also a matrix

MNIST contains 28×28-pixel handwritten digits. One pixel is a scalar; one image is a matrix; many images form a tensor.

```python
from tensorflow.keras.datasets import mnist
(x_mnist, y_mnist), _ = mnist.load_data()
print("Dataset:", x_mnist.shape)
print("Label:", y_mnist[0])
print("One image:", x_mnist[0].shape)
```

```python
plt.figure(figsize=(3,3))
plt.imshow(x_mnist[0], cmap="gray")
plt.title(f"Label: {y_mnist[0]}")
plt.axis("off")
plt.show()
```

```python
print("Scalar (one pixel):", x_mnist[0,14,14])
print("Vector (one row):", x_mnist[0,14].shape)
print("Matrix (one image):", x_mnist[0].shape)
print("Tensor (dataset):", x_mnist.shape)
print("Rank/ndim:", x_mnist.ndim)
```

## 3. IMDB: text must be converted into numbers

Words are transformed into token identifiers. The sequences are then padded to a fixed length to form a matrix/tensor.

```python
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
(x_imdb, y_imdb), _ = imdb.load_data(num_words=10_000)
print("First tokens:", x_imdb[0][:20])
print("Sentiment (0/1):", y_imdb[0])
x_imdb_pad = pad_sequences(x_imdb, maxlen=200)
print("Tensor IMDB:", x_imdb_pad.shape)
```

## 4. ImageNet: RGB image and batch dimension

Downloading the original ImageNet is impractical in a live session. We use an RGB image in the format consumed by ImageNet-pretrained models. The notebook also includes an **optional** ImageNet-v2 cell (large download).

```python
import tensorflow as tf
url = "https://storage.googleapis.com/download.tensorflow.org/example_images/YellowLabradorLooking_new.jpg"
path = tf.keras.utils.get_file("dog.jpg", origin=url)
img = tf.keras.utils.load_img(path, target_size=(224,224))
img_np = tf.keras.utils.img_to_array(img)
batch = np.expand_dims(img_np, axis=0)
print("One RGB image:", img_np.shape)
print("Batch of 1 image:", batch.shape)
```

```python
plt.figure(figsize=(4,4))
plt.imshow(img_np.astype("uint8"))
plt.axis("off")
plt.show()
```

## 5. The same tensor in three ecosystems

```python
import tensorflow as tf
import torch
A_np = np.array([[1.,2.],[3.,4.]])
A_tf = tf.constant(A_np)
A_torch = torch.tensor(A_np)
print("NumPy     ", type(A_np), A_np.shape)
print("TensorFlow", type(A_tf), A_tf.shape)
print("PyTorch   ", type(A_torch), tuple(A_torch.shape))
```

## 6. Broadcasting

A common deep learning operation: adding a vector to every row of a matrix without copying it manually.

```python
A = np.array([[1,2,3],[4,5,6]])
b = np.array([10,20,30])
print(A + b)
```
