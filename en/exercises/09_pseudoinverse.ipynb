```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
print("NumPy:", np.__version__)
```

## When a regular inverse does not apply

The pseudoinverse generalizes the idea to rectangular matrices or systems without a unique exact solution.

```python
A=np.array([[1.,0.],[1.,1.],[1.,2.],[1.,3.]])
y=np.array([1.0,2.2,2.8,4.1])
A_plus=np.linalg.pinv(A)
beta=A_plus@y
print("Coeficientes [intercepto, pendiente]:",beta)
print("Predicciones:",A@beta)
```

## Relationship to least squares

```python
beta_lstsq,*_=np.linalg.lstsq(A,y,rcond=None)
print("pinv:",beta)
print("lstsq:",beta_lstsq)
print("Coinciden:",np.allclose(beta,beta_lstsq))
```

```python
x=np.linspace(0,3,100)
plt.scatter(A[:,1],y,label='data')
plt.plot(x,beta[0]+beta[1]*x,label='fit')
plt.legend(); plt.title('Linear regression via pseudoinverse'); plt.show()
```

### Connection to ML

The pseudoinverse produces the least-squares solution to linear problems and clarifies what ‘best approximation’ means when no exact solution exists.
