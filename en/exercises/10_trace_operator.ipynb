## Trace = sum of the diagonal

```python
A=np.array([[2.,1.,0.],[4.,3.,5.],[1.,2.,7.]])
print("Trace:",np.trace(A))
print("Diagonal sum:",np.diag(A).sum())
```

## Identity involving the Frobenius norm

```python
left=np.linalg.norm(A,'fro')
right=np.sqrt(np.trace(A@A.T))
print(left,right,np.allclose(left,right))
```

## Cyclic invariance

```python
B=np.random.randn(3,4); C=np.random.randn(4,3)
print("Tr(BC)=",np.trace(B@C))
print("Tr(CB)=",np.trace(C@B))
```

### Connection to ML

The trace expresses variance, covariance, and norms compactly, which is common in optimization and probability derivations.
