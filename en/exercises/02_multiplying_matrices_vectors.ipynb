## Dot product: combining features

An artificial neuron takes inputs, multiplies them by weights, and adds the results.

```python
x = np.array([0.8, 0.3, 0.5])
w = np.array([1.2, -0.7, 0.4])
score = x @ w
print("x · w =", score)
```

## Matrix-vector product

Many observations can be processed in a single operation.

```python
X = np.array([[0.8,0.3,0.5],[0.2,0.9,0.1],[0.5,0.5,0.5]])
y = X @ w
print("X shape:", X.shape, "w shape:", w.shape, "result:", y.shape)
print(y)
```

## Matrix-matrix product: a dense layer

If `X` contains a batch and `W` contains several sets of weights, `X @ W` produces several outputs per example.

```python
W = np.random.randn(3,4)
Z = X @ W
print("X:", X.shape, "W:", W.shape, "Z:", Z.shape)
print(Z)
```

## With MNIST: 28×28 → 784 → 10 scores

```python
from tensorflow.keras.datasets import mnist
(x_train, y_train), _ = mnist.load_data()
X_batch = x_train[:32].reshape(32,-1).astype("float32")/255.0
W_digits = np.random.randn(784,10)*0.01
logits = X_batch @ W_digits
print("Batch:", X_batch.shape)
print("Weights:", W_digits.shape)
print("Logits:", logits.shape)
```

## NumPy vs TensorFlow vs PyTorch

```python
import tensorflow as tf, torch
A = np.random.randn(128,64).astype("float32")
B = np.random.randn(64,32).astype("float32")
print("NumPy:", (A @ B).shape)
print("TensorFlow:", tf.matmul(A,B).shape)
print("PyTorch:", tuple((torch.from_numpy(A) @ torch.from_numpy(B)).shape))
```

### Key idea

Neural networks may seem complex, but much of their internal computation reduces to **matrix multiplications + nonlinear functions**.
