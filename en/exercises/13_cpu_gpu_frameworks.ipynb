## Goal

Compare the same matrix operation in three frameworks. We are not seeking a scientific benchmark; we want to observe **device, shape, and transfer**.

```python
import time, numpy as np, tensorflow as tf, torch
N=1500
A=np.random.randn(N,N).astype('float32')
B=np.random.randn(N,N).astype('float32')
print("TF GPU:",tf.config.list_physical_devices('GPU'))
print("Torch CUDA:",torch.cuda.is_available())
```

## NumPy — CPU

```python
t=time.perf_counter(); C=A@B; dt=time.perf_counter()-t
print(f"NumPy CPU: {dt:.3f}s | {C.shape}")
```

## TensorFlow — the runtime selects the available CPU/GPU

```python
A_tf=tf.constant(A); B_tf=tf.constant(B)
t=time.perf_counter(); C_tf=tf.matmul(A_tf,B_tf); _=C_tf.numpy(); dt=time.perf_counter()-t
print(f"TensorFlow: {dt:.3f}s | {C_tf.shape} | {C_tf.device}")
```

## PyTorch — move explicitly to the device

```python
device='cuda' if torch.cuda.is_available() else 'cpu'
A_t=torch.from_numpy(A).to(device); B_t=torch.from_numpy(B).to(device)
if device=='cuda': torch.cuda.synchronize()
t=time.perf_counter(); C_t=A_t@B_t
if device=='cuda': torch.cuda.synchronize()
dt=time.perf_counter()-t
print(f"PyTorch {device}: {dt:.3f}s | {tuple(C_t.shape)}")
```

### Class takeaway

The algebra is the same. What changes across frameworks is the API, device management, autograd, and the ecosystem. GPUs are especially efficient at running many tensor operations in parallel.
