```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
print("NumPy:", np.__version__)
```

## Intuition

An eigenvector keeps its direction under a linear transformation; its eigenvalue indicates how much it is scaled.

```python
A=np.array([[3.,1.],[0.,2.]])
vals, vecs=np.linalg.eig(A)
print("Autovalores:",vals)
print("Autovectores (columnas):")
print(vecs)
for i in range(2):
    v=vecs[:,i]
    print(np.allclose(A@v, vals[i]*v))
```

## In data: principal directions of variation

```python
X=np.random.multivariate_normal([0,0], [[3,2.2],[2.2,2]], size=500)
C=np.cov(X,rowvar=False)
vals,vecs=np.linalg.eigh(C)
order=np.argsort(vals)[::-1]; vals=vals[order]; vecs=vecs[:,order]
print("Covarianza:")
print(C)
print("Autovalores:",vals)
```

```python
plt.figure(figsize=(6,5))
plt.scatter(X[:,0],X[:,1],s=10,alpha=.45)
center=X.mean(0)
for i in range(2):
    v=vecs[:,i]*np.sqrt(vals[i])*2
    plt.arrow(center[0],center[1],v[0],v[1],width=.03,length_includes_head=True)
plt.axis('equal'); plt.title('Eigenvectors of the covariance matrix'); plt.show()
```

### Connection to ML

This idea is the geometric heart of PCA: finding the directions along which data varies most.
