## Cuando una inversa normal no aplica

La pseudoinversa generaliza la idea a matrices rectangulares o sistemas sin solución exacta única.

```python
A=np.array([[1.,0.],[1.,1.],[1.,2.],[1.,3.]])
y=np.array([1.0,2.2,2.8,4.1])
A_plus=np.linalg.pinv(A)
beta=A_plus@y
print("Coeficientes [intercepto, pendiente]:",beta)
print("Predicciones:",A@beta)
```

## Relación con mínimos cuadrados

```python
beta_lstsq,*_=np.linalg.lstsq(A,y,rcond=None)
print("pinv:",beta)
print("lstsq:",beta_lstsq)
print("Coinciden:",np.allclose(beta,beta_lstsq))
```

```python
x=np.linspace(0,3,100)
plt.scatter(A[:,1],y,label='datos')
plt.plot(x,beta[0]+beta[1]*x,label='ajuste')
plt.legend(); plt.title('Regresión lineal vía pseudoinversa'); plt.show()
```

### Conexión con ML

La pseudoinversa produce la solución de mínimos cuadrados en problemas lineales y ayuda a entender qué significa “mejor aproximación” cuando no existe una solución exacta.
