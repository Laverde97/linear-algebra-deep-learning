## Diagonal e identidad

```python
d=np.array([2.,5.,9.])
D=np.diag(d)
I=np.eye(3)
print("Diagonal:")
print(D)
print("Identidad:")
print(I)
```

## Matriz simétrica

```python
A=np.array([[2.,1.,0.],[1.,3.,4.],[0.,4.,5.]])
print("A == A.T ?", np.allclose(A,A.T))
```

## Vectores unitarios y ortogonales

```python
x=np.array([3.,4.])
x_unit=x/np.linalg.norm(x)
y=np.array([-4.,3.]); y_unit=y/np.linalg.norm(y)
print("||x_unit||=",np.linalg.norm(x_unit))
print("x_unit · y_unit=",x_unit@y_unit)
```

## One-hot: un vector especial muy usado en clasificación

```python
label=7
one_hot=np.eye(10)[label]
print(one_hot)
```

### Conexión con IA

Matrices diagonales simplifican transformaciones, matrices simétricas aparecen en covarianzas y vectores one-hot representan categorías.
