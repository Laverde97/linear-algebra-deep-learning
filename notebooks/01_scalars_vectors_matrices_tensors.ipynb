## 1. Datos aleatorios simulados

Empezamos con información tabular. Un solo valor es un **escalar**, una fila es un **vector**, toda la tabla es una **matriz**.

```python
edades = np.random.randint(18, 80, 100)
peso = np.random.uniform(50, 100, 100)
altura = np.random.uniform(1.50, 1.95, 100)
ingresos = np.random.uniform(1_000_000, 12_000_000, 100)

datos = np.column_stack([edades, peso, altura, ingresos])
print("Escalar:", datos[0,0])
print("Vector (una persona):", datos[0].shape)
print("Matriz (100 personas):", datos.shape)
print(datos[:3])
```

## 2. MNIST: una imagen también es una matriz

MNIST tiene dígitos manuscritos de 28×28 píxeles. Un píxel es un escalar; una imagen es una matriz; muchas imágenes forman un tensor.

```python
from tensorflow.keras.datasets import mnist
(x_mnist, y_mnist), _ = mnist.load_data()
print("Dataset:", x_mnist.shape)
print("Etiqueta:", y_mnist[0])
print("Una imagen:", x_mnist[0].shape)
```

```python
plt.figure(figsize=(3,3))
plt.imshow(x_mnist[0], cmap="gray")
plt.title(f"Etiqueta: {y_mnist[0]}")
plt.axis("off")
plt.show()
```

```python
print("Escalar (un píxel):", x_mnist[0,14,14])
print("Vector (una fila):", x_mnist[0,14].shape)
print("Matriz (una imagen):", x_mnist[0].shape)
print("Tensor (dataset):", x_mnist.shape)
print("Rango/ndim:", x_mnist.ndim)
```

## 3. IMDB: el texto debe convertirse en números

Las palabras se transforman en identificadores de tokens. Luego las secuencias se igualan a una longitud fija para formar una matriz/tensor.

```python
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
(x_imdb, y_imdb), _ = imdb.load_data(num_words=10_000)
print("Primeros tokens:", x_imdb[0][:20])
print("Sentimiento (0/1):", y_imdb[0])
x_imdb_pad = pad_sequences(x_imdb, maxlen=200)
print("Tensor IMDB:", x_imdb_pad.shape)
```

## 4. ImageNet: imagen RGB y dimensión batch

En una práctica en vivo no conviene descargar el ImageNet original. Usamos una imagen RGB en el formato que consumen modelos preentrenados en ImageNet. El notebook deja también una celda **opcional** para ImageNet-v2 (descarga grande).

```python
import tensorflow as tf
url = "https://storage.googleapis.com/download.tensorflow.org/example_images/YellowLabradorLooking_new.jpg"
path = tf.keras.utils.get_file("dog.jpg", origin=url)
img = tf.keras.utils.load_img(path, target_size=(224,224))
img_np = tf.keras.utils.img_to_array(img)
batch = np.expand_dims(img_np, axis=0)
print("Una imagen RGB:", img_np.shape)
print("Batch de 1 imagen:", batch.shape)
```

```python
plt.figure(figsize=(4,4))
plt.imshow(img_np.astype("uint8"))
plt.axis("off")
plt.show()
```

## 5. El mismo tensor en tres ecosistemas

```python
import tensorflow as tf
import torch
A_np = np.array([[1.,2.],[3.,4.]])
A_tf = tf.constant(A_np)
A_torch = torch.tensor(A_np)
print("NumPy     ", type(A_np), A_np.shape)
print("TensorFlow", type(A_tf), A_tf.shape)
print("PyTorch   ", type(A_torch), tuple(A_torch.shape))
```

## 6. Broadcasting

Una operación muy común en deep learning: sumar un vector a cada fila de una matriz sin copiarlo manualmente.

```python
A = np.array([[1,2,3],[4,5,6]])
b = np.array([10,20,30])
print(A + b)
```
