## Dependencia lineal

Un vector es redundante si puede construirse como combinación de otros.

```python
v1 = np.array([1.,2.])
v2 = np.array([2.,4.])  # 2*v1
A = np.column_stack([v1,v2])
print(A)
print("Rango:", np.linalg.matrix_rank(A))
```

## Span: todas las combinaciones posibles

```python
v1 = np.array([1.,0.])
v2 = np.array([0.5,1.])
pts=[]
for a in np.linspace(-2,2,15):
    for b in np.linspace(-2,2,15):
        pts.append(a*v1+b*v2)
pts=np.array(pts)
plt.figure(figsize=(6,5))
plt.scatter(pts[:,0],pts[:,1],s=12)
plt.axhline(0,linewidth=.8); plt.axvline(0,linewidth=.8)
plt.title("Combinaciones lineales de dos vectores independientes")
plt.show()
```

## Redundancia en un dataset simulado

```python
x1 = np.random.randn(200)
x2 = 3*x1  # característica completamente redundante
x3 = np.random.randn(200)
X = np.column_stack([x1,x2,x3])
print("Columnas:", X.shape[1])
print("Rango real:", np.linalg.matrix_rank(X))
```

### Conexión con ML

Variables redundantes pueden generar multicolinealidad, inestabilidad y cálculos innecesarios. El rango ayuda a saber cuánta información lineal independiente existe.
