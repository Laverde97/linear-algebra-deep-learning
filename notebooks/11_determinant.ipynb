```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
print("NumPy:", np.__version__)
```

## Determinante como cambio de área/volumen

```python
A=np.array([[2.,0.5],[0.2,1.5]])
print("det(A)=",np.linalg.det(A))
```

```python
square=np.array([[0,0],[1,0],[1,1],[0,1],[0,0]],dtype=float)
trans=square@A.T
plt.figure(figsize=(6,5))
plt.plot(square[:,0],square[:,1],marker='o',label='original')
plt.plot(trans[:,0],trans[:,1],marker='o',label='transformado')
plt.axis('equal'); plt.legend(); plt.title(f'Área escalada por |det|={abs(np.linalg.det(A)):.2f}'); plt.show()
```

## Determinante cero = colapso de dimensión

```python
S=np.array([[1.,2.],[2.,4.]])
print("det(S)=",np.linalg.det(S))
print("rango=",np.linalg.matrix_rank(S))
```

### Conexión con ML

Determinantes aparecen en cambios de variables, densidades gaussianas y análisis de transformaciones. Un determinante cero revela pérdida de una dimensión.
